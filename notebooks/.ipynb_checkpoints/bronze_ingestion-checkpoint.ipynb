```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, from_json, current_timestamp

#1. khoi tao sparkSession
spark = SparkSession.builder.appName("bronze_Ingestion").getOrCreate()

#2. doc luong data real time tu kafka
df_kafka = spark.readStream \
    .format("kafka") \
    #port mac dinh cua redpanda
    .option("kafka.bootstrap.servers","redpanda:9092") \ 
    #dang ky topic da config
    .option("subscribe","yellow_trips") \
    .option("startingOffsets", "earliest") \
    .load()


df_yellow_nyc = (
    df_kafka \
    .withColumn("json_str", col("value").cast("string")) \
    .withColumn("ingested_at", current_timestamp()) \
    .select(
        from_json(col("json_str")).alias("data"),
        col("offset"),
    ) \
    .select(
        "offset", 
        "ingested_at", 
        "data.*"
    )          
            
)


query = df_yellow_nyc.writeStream \
    .format("iceberg") \
    .outputMode("append") \
    .option("checkpointLocation","/data/checkpoints/bronze_trips") \
    .toTable("bronze.taxi_trips_raw") \
    .awaitTermination() \

    

```
