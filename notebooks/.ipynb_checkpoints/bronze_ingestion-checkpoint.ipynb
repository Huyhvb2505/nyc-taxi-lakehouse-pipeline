```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, current_timestamp

# 1. Khởi tạo Spark Session & iceberg
spark = SparkSession.builder \
    .appName("Bronze_Ingestion") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.iceberg", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.iceberg.type", "hadoop") \
    .config("spark.sql.catalog.iceberg.warehouse", "file:///data/lakehouse/iceberg") \
    .getOrCreate()

# create namespace on catalog iceberg
spark.sql("CREATE NAMESPACE IF NOT EXISTS iceberg.bronze")


# 2. real data real-time from kafka(Redpanda)
df_kafka = spark.readStream \
    .format("kafka") \
    .option("kafka.bootstrap.servers", "redpanda:9092") \
    .option("subscribe", "yellow_trips") \
    .option("startingOffsets", "earliest") \
    .option("maxOffsetsPerTrigger", "2000") \
    .load()

# 3. transform data Bronze layer
df_bronze = df_kafka.select(
    col("offset"),
    col("partition"),
    col("value").cast("string").alias("json_str"),
    current_timestamp().alias("ingested_at")
)

# 4. write stream into warehouse(iceberg table)
# checkpoint, append mod => No data loss
query = df_bronze.writeStream \
    .format("iceberg") \
    .outputMode("append") \
    .option("checkpointLocation", "/data/checkpoints/bronze_trips") \
    .toTable("iceberg.bronze.taxi_trips_raw")

# Kích hoạt luồng chạy liên tục
query.awaitTermination()
```
