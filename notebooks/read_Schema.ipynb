```python
from pyspark.sql import SparkSession

# 1. Khởi tạo Spark Session (Nếu bạn chưa khởi tạo ở cell trước đó)
spark = SparkSession.builder.appName("Explore_Data").getOrCreate()

# 2. Đọc dữ liệu từ file Parquet (Ví dụ: tháng 01/2024)
df_yellow = spark.read.parquet("/data/raw/yellow/yellow_tripdata_2024-01.parquet")

# 3. --- XEM SCHEMA ---
print("=== CẤU TRÚC BẢNG (SCHEMA) ===")
df_yellow.printSchema()

# 4. --- XEM DATA ---
print("=== DỮ LIỆU 5 DÒNG ĐẦU TIÊN ===")
# Sử dụng toPandas() để hiển thị giao diện bảng đẹp hơn trên Jupyter
display(df_yellow.limit(5).toPandas())
```
